```python
contact_data = pd.read_csv ('contactsshort.csv')
contact_data.head()
```

```python
#fill null with 'na' value
contact_data = contact_data.fillna('na')
contact_data.head()
```

```python
#replacing x value for gender with yes or no
contact_data['Man'] = contact_data['Man'].str.replace('x','yes')
contact_data['Man'] = contact_data['Man'].str.replace('na','no')
contact_data['Woman'] = contact_data['Woman'].str.replace('x','yes')
contact_data['Woman'] = contact_data['Woman'].str.replace('na','no')
contact_data['Nonbinary'] = contact_data['Nonbinary'].str.replace('x','yes')
contact_data['Nonbinary'] = contact_data['Nonbinary'].str.replace('na','no')
contact_data['Notlisted'] = contact_data['Notlisted'].str.replace('x','yes')
contact_data['Notlisted'] = contact_data['Notlisted'].str.replace('na','no')
contact_data['Prefnoanswer'] = contact_data['Prefnoanswer'].str.replace('x','yes')
contact_data['Prefnoanswer'] = contact_data['Prefnoanswer'].str.replace('na','no')
contact_data.head()
```

```python
contact_data.columns
```

```python
contact_data['Gender'] = None
contact_data['State_value'] = None
contact_data['gender_value'] = None
```

```python
contact_data.columns
```

```python
# look for distinct values in state columns to make sure they all have an assigned value
```

```python
state_num = {state: i + 1 for i, state in enumerate(['AL', 'AK', 'AZ', 'AR', 'CA', 'CO', 'CT', 'DE', 'FL', 'GA', 'HI', 'ID', 'IL', 'IN', 'IA', 'KS', 'KY', 'LA', 'ME', 'MD', 'MA', 'MI', 'MN', 'MS', 'MO', 'MT', 'NE', 'NV', 'NH', 'NJ', 'NM', 'NY', 'NC', 'ND', 'OH', 'OK', 'OR', 'PA', 'RI', 'SC', 'SD', 'TN', 'TX', 'UT', 'VT', 'VA', 'WA', 'WV', 'WI', 'WY', 'DC'])}
contact_data.head()
```

```python
contact_data['State_value'] = contact_data['pState/Province'].map(state_num)
contact_data.head()
```

```python
for index, row in contact_data.iterrows():
    if row['Man'] == 'yes':
        contact_data.loc[index, 'Gender'] = 'Man'
    elif row['Woman'] == 'yes':
        contact_data.loc[index, 'Gender'] = 'Woman'
    elif row['Nonbinary'] == 'yes':
        contact_data.loc[index, 'Gender'] = 'Nonbinary'
    elif row['Notlisted'] == 'yes':
        contact_data.loc[index, 'Gender'] = 'Notlisted'
    elif row['Prefnoanswer'] == 'yes':
        contact_data.loc[index, 'Gender'] = 'Prefnoanswer'
```

```python
#data dictionary for gender: 1=man, 2=woman, 3=nonbinary, 4=notlisted, 5=prefnoanswer
gender_num = {gender: i + 1 for i, gender in enumerate(['Man', 'Woman', 'Nonbinary','Notlisted','Prefnoanswer'])}
contact_data['gender_value'] = contact_data['Gender'].map(gender_num)
contact_data.head()
```

```python
contact_data.head()
```

```python
#dropping gender columns and other columns we are not using for the model
contact_data = contact_data.drop(columns=['Man', 
                                          'Woman', 
                                          'Nonbinary', 
                                          'Notlisted', 
                                          'Prefnoanswer',
                                          'VANID', 
                                          'Engagement Points As Of Today', 
                                          'Engagement Percentile As Of Today', 
                                          'Engagement Points As Of 7 Days Ago', 
                                          'Engagement Percentile As Of 7 Days Ago',
                                          'Engagement Points As Of 30 Days Ago', 
                                          'Engagement Percentile As Of 30 Days Ago', 
                                          'Engagement Points As Of 1 Year Ago',
                                          'Engagement Percentile As Of 1 Year Ago',
                                          'Engagement Points As Of 2 Years Ago',
                                          'Engagement Percentile As Of 2 Years Ago', 
                                          'pAddress', 
                                          'pCity', 
                                          'pZip/Postal', 
                                          'pZip4', 
                                          'pCountryCode',
                                          'pCountryName', 
                                          'pAddressID'
                                         ])
contact_data.head()
```
